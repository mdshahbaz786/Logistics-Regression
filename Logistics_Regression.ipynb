1.Question 1: What is Simple Linear Regression (SLR)? Explain its purpose.

**Answer:**

Simple Linear Regression (SLR) is a supervised machine learning algorithm used to find and model the relationship between **one independent variable (input)** and **one dependent variable (output)**. It represents this relationship using a straight line.

The main purpose of Simple Linear Regression is to **predict the value of the dependent variable based on the value of the independent variable**.

For example,- SLR can be used to predict a student's exam score based on the number of hours they studied. Here, the number of study hours is the independent variable, and the exam score is the dependent variable.

2. What are the key assumptions of Simple Linear Regression?  
Answer:

The key assumptions of Simple Linear Regression are:

i. Linearity: There should be a linear relationship between the independent variable and the dependent variable.

ii. Independence: The observations and errors should be independent of each other.
iii. Homoscedasticity: The variance of the errors should remain approximately constant for all values of the independent variable.

iv. Normality of errors: The residuals (errors) should be approximately normally distributed.

v. No significant outliers: The data should not contain extreme outliers that can strongly influence the regression model.

Question 3: Write the mathematical equation for a simple linear regression model and explain each term.

**Answer:**

The mathematical equation for a Simple Linear Regression model is:

**y = β₀ + β₁x + ε**

Where:

* **y** = Dependent variable or the value we want to predict.
* **x** = Independent variable used to make the prediction.
* **β₀** = Intercept, which represents the expected value of y when x is 0.
* **β₁** = Slope or regression coefficient, which represents the change in y for a one-unit increase in x.
* **ε** = Error term, which represents the difference between the actual value and the predicted value.

The prediction equation can also be written as:

**ŷ = b₀ + b₁x**

This equation represents the straight line used by the regression model to make predictions.

Question 5: What is the method of least squares in linear regression?

**Answer:**

The **method of least squares** is a technique used in linear regression to find the best-fitting straight line for a given set of data.

It calculates the difference between the actual values and the predicted values. These differences are called **errors or residuals**. The method squares each error and finds the regression line that minimizes the sum of all squared errors.

The objective is:

**Minimize Σ(yᵢ − ŷᵢ)²**

Where:

* **yᵢ** = Actual value
* **ŷᵢ** = Predicted value
* **yᵢ − ŷᵢ** = Error or residual

Therefore, the least squares method chooses the line that gives the smallest possible total squared error.

Question 6: What is Logistic Regression? How does it differ from Linear Regression?

**Answer:**

**Logistic Regression** is a supervised machine learning algorithm mainly used for **classification problems**. It predicts the probability that an observation belongs to a particular class.

For example, Logistic Regression can be used to predict whether an email is **spam or not spam**.

The main differences are:

| Linear Regression                                       | Logistic Regression                                      |
| ------------------------------------------------------- | -------------------------------------------------------- |
| Used mainly for predicting continuous numerical values. | Used mainly for classification problems.                 |
| Produces continuous output values.                      | Produces probabilities between 0 and 1.                  |
| Uses a linear relationship to predict the output.       | Uses the logistic/sigmoid function to model probability. |
| Example: predicting house prices.                       | Example: predicting whether an email is spam.            |

Therefore, Linear Regression is generally used for **regression problems**, while Logistic Regression is generally used for **classification problems**.

Question 7: Name and briefly describe three common evaluation metrics for regression models.

**Answer:**

Three common evaluation metrics for regression models are:

**1. Mean Absolute Error (MAE):**

MAE calculates the average absolute difference between the actual and predicted values.

**MAE = (1/n) Σ|yᵢ − ŷᵢ|**

A lower MAE indicates better model performance.

**2. Mean Squared Error (MSE):**

MSE calculates the average of the squared differences between actual and predicted values.

**MSE = (1/n) Σ(yᵢ − ŷᵢ)²**

MSE gives more importance to larger errors because the errors are squared. A lower MSE indicates better performance.

**3. R-squared (R²):**

R² measures how much of the variation in the dependent variable is explained by the regression model. A higher R² generally indicates a better fit to the observed data.

Question 8: What is the purpose of the R-squared metric in regression analysis?

**Answer:**

The purpose of **R-squared (R²)** is to measure how well a regression model explains the variation in the dependent variable.

For example, an R² value of **0.80** means that approximately **80% of the variation** in the dependent variable is explained by the model on the evaluated data.

The formula is:

**R² = 1 − (SSres / SStot)**

Where:

* **SSres** = Sum of squared residuals.
* **SStot** = Total sum of squares.

A higher R² generally indicates that the model explains more of the observed variation. However, R² alone does not prove that the model is correct or that there is a causal relationship between the variables.

Question 10: How do you interpret the coefficients in a simple linear regression model?

**Answer:**

In a Simple Linear Regression model:

**ŷ = b₀ + b₁x**

the coefficients are interpreted as follows:

* **Intercept (b₀):** It represents the predicted value of the dependent variable when the independent variable is zero.
* **Slope (b₁):** It represents the expected change in the dependent variable for every one-unit increase in the independent variable.

For example, suppose a model for predicting salary based on years of experience is:

**Salary = 30,000 + 5,000 × Experience**

Here, **30,000** is the intercept, meaning the predicted salary when experience is 0 years.

**5,000** is the slope, meaning that each additional year of experience is associated with an estimated increase of 5,000 in salary, assuming the model is appropriate.

Therefore, the coefficients help us understand the baseline value and the direction and size of the relationship between the variables.



```python
#Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.

from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [2, 4, 6, 8, 10]

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```
